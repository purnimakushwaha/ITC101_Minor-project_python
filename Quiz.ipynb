```python
# ============================================================
#              SMART QUIZ & EXAM MANAGEMENT SYSTEM
# ============================================================

import pandas as pd
import random
import time
import os
from datetime import datetime


# ============================================================
# PROJECT SETTINGS
# ============================================================

RESULT_FILE = "quiz_results.csv"

quiz_results = []


# ============================================================
# QUESTION BANK
# ============================================================

questions = {

    "Python": [

        {
            "question": "Which keyword is used to define a function in Python?",
            "options": ["A. function", "B. def", "C. fun", "D. define"],
            "answer": "B"
        },

        {
            "question": "Which data type is used to store True or False?",
            "options": ["A. int", "B. str", "C. bool", "D. float"],
            "answer": "C"
        },

        {
            "question": "Which symbol is used for comments in Python?",
            "options": ["A. //", "B. <!--", "C. #", "D. **"],
            "answer": "C"
        },

        {
            "question": "Which function is used to display output in Python?",
            "options": ["A. display()", "B. print()", "C. output()", "D. show()"],
            "answer": "B"
        },

        {
            "question": "Which keyword is used to create a loop over a sequence?",
            "options": ["A. repeat", "B. loop", "C. for", "D. foreach"],
            "answer": "C"
        },

        {
            "question": "Which collection is ordered and changeable?",
            "options": ["A. List", "B. Tuple", "C. Set", "D. FrozenSet"],
            "answer": "A"
        },

        {
            "question": "What is the extension of a Python file?",
            "options": ["A. .java", "B. .py", "C. .python", "D. .pt"],
            "answer": "B"
        },

        {
            "question": "Which function returns the length of an object?",
            "options": ["A. count()", "B. size()", "C. len()", "D. length()"],
            "answer": "C"
        },

        {
            "question": "Which operator is used for exponentiation?",
            "options": ["A. ^", "B. **", "C. //", "D. %%"],
            "answer": "B"
        },

        {
            "question": "Which keyword is used to handle exceptions?",
            "options": ["A. error", "B. catch", "C. try", "D. exception"],
            "answer": "C"
        }
    ],


    "Computer": [

        {
            "question": "What does CPU stand for?",
            "options": [
                "A. Central Processing Unit",
                "B. Computer Processing Unit",
                "C. Central Program Unit",
                "D. Computer Program Utility"
            ],
            "answer": "A"
        },

        {
            "question": "Which device is used to input text?",
            "options": [
                "A. Monitor",
                "B. Keyboard",
                "C. Speaker",
                "D. Projector"
            ],
            "answer": "B"
        },

        {
            "question": "Which one is an operating system?",
            "options": [
                "A. Python",
                "B. Windows",
                "C. HTML",
                "D. SQL"
            ],
            "answer": "B"
        },

        {
            "question": "What does RAM stand for?",
            "options": [
                "A. Read Access Memory",
                "B. Random Access Memory",
                "C. Rapid Access Machine",
                "D. Run Access Memory"
            ],
            "answer": "B"
        },

        {
            "question": "Which device is used to display output?",
            "options": [
                "A. Keyboard",
                "B. Mouse",
                "C. Monitor",
                "D. Scanner"
            ],
            "answer": "C"
        },

        {
            "question": "Which is a web browser?",
            "options": [
                "A. Chrome",
                "B. Windows",
                "C. Python",
                "D. Linux"
            ],
            "answer": "A"
        },

        {
            "question": "What does USB stand for?",
            "options": [
                "A. Universal Serial Bus",
                "B. United System Bus",
                "C. Universal System Board",
                "D. User Serial Board"
            ],
            "answer": "A"
        },

        {
            "question": "Which one is a programming language?",
            "options": [
                "A. Python",
                "B. Google",
                "C. Windows",
                "D. Keyboard"
            ],
            "answer": "A"
        },

        {
            "question": "Which memory is temporary?",
            "options": [
                "A. ROM",
                "B. Hard Disk",
                "C. RAM",
                "D. SSD"
            ],
            "answer": "C"
        },

        {
            "question": "Which device is used to print documents?",
            "options": [
                "A. Scanner",
                "B. Printer",
                "C. Monitor",
                "D. Keyboard"
            ],
            "answer": "B"
        }
    ],


    "General Knowledge": [

        {
            "question": "What is the capital of India?",
            "options": [
                "A. Mumbai",
                "B. New Delhi",
                "C. Kolkata",
                "D. Chennai"
            ],
            "answer": "B"
        },

        {
            "question": "Which planet is known as the Red Planet?",
            "options": [
                "A. Earth",
                "B. Venus",
                "C. Mars",
                "D. Jupiter"
            ],
            "answer": "C"
        },

        {
            "question": "How many days are there in a leap year?",
            "options": [
                "A. 365",
                "B. 366",
                "C. 364",
                "D. 367"
            ],
            "answer": "B"
        },

        {
            "question": "Which is the largest ocean?",
            "options": [
                "A. Atlantic Ocean",
                "B. Indian Ocean",
                "C. Arctic Ocean",
                "D. Pacific Ocean"
            ],
            "answer": "D"
        },

        {
            "question": "How many continents are there?",
            "options": [
                "A. 5",
                "B. 6",
                "C. 7",
                "D. 8"
            ],
            "answer": "C"
        },

        {
            "question": "Which gas do plants mainly absorb?",
            "options": [
                "A. Oxygen",
                "B. Carbon Dioxide",
                "C. Nitrogen",
                "D. Hydrogen"
            ],
            "answer": "B"
        },

        {
            "question": "Which is the largest planet in our solar system?",
            "options": [
                "A. Earth",
                "B. Saturn",
                "C. Jupiter",
                "D. Neptune"
            ],
            "answer": "C"
        },

        {
            "question": "How many hours are there in one day?",
            "options": [
                "A. 12",
                "B. 18",
                "C. 24",
                "D. 48"
            ],
            "answer": "C"
        },

        {
            "question": "Which is the fastest land animal?",
            "options": [
                "A. Lion",
                "B. Tiger",
                "C. Cheetah",
                "D. Horse"
            ],
            "answer": "C"
        },

        {
            "question": "Which instrument is used to measure temperature?",
            "options": [
                "A. Barometer",
                "B. Thermometer",
                "C. Speedometer",
                "D. Altimeter"
            ],
            "answer": "B"
        }
    ]
}


# ============================================================
# GRADE FUNCTION
# ============================================================

def calculate_grade(percentage):

    if percentage >= 90:
        return "A+"

    elif percentage >= 80:
        return "A"

    elif percentage >= 70:
        return "B+"

    elif percentage >= 60:
        return "B"

    elif percentage >= 50:
        return "C"

    elif percentage >= 40:
        return "D"

    else:
        return "F"


# ============================================================
# START QUIZ
# ============================================================

def start_quiz():

    print("\n" + "=" * 60)
    print("                    START QUIZ")
    print("=" * 60)

    student_name = input(
        "Enter student name: "
    ).strip()

    if student_name == "":
        print("Name cannot be empty.")
        return

    # Category selection
    print("\nSelect Quiz Category:")

    categories = list(
        questions.keys()
    )

    for i, category in enumerate(
        categories,
        start=1
    ):
        print(
            f"{i}. {category}"
        )

    try:

        category_choice = int(
            input(
                "Enter category number: "
            )
        )

        if not 1 <= category_choice <= len(categories):

            print("Invalid category.")

            return

    except ValueError:

        print(
            "Please enter a valid number."
        )

        return

    category = categories[
        category_choice - 1
    ]

    # Number of questions
    available_questions = questions[
        category
    ]

    print(
        f"\nCategory selected: {category}"
    )

    print(
        f"Available questions: {len(available_questions)}"
    )

    try:

        number_of_questions = int(
            input(
                "How many questions? "
            )
        )

        if (
            number_of_questions < 1
            or number_of_questions > len(available_questions)
        ):

            print(
                f"Choose between 1 and "
                f"{len(available_questions)}."
            )

            return

    except ValueError:

        print(
            "Please enter a valid number."
        )

        return

    # Random questions
    selected_questions = random.sample(
        available_questions,
        number_of_questions
    )

    print("\n" + "=" * 60)
    print("                    EXAM STARTED")
    print("=" * 60)

    print(
        f"Student   : {student_name}"
    )

    print(
        f"Category  : {category}"
    )

    print(
        f"Questions : {number_of_questions}"
    )

    print("=" * 60)

    input(
        "\nPress ENTER to start the exam..."
    )

    # Start timer
    start_time = time.perf_counter()

    start_datetime = datetime.now()

    score = 0

    correct_answers = []

    wrong_answers = []

    # ========================================================
    # QUESTION LOOP
    # ========================================================

    for question_number, question_data in enumerate(
        selected_questions,
        start=1
    ):

        print("\n" + "-" * 60)

        print(
            f"Question {question_number} "
            f"of {number_of_questions}"
        )

        print("-" * 60)

        print(
            question_data["question"]
        )

        print()

        for option in question_data["options"]:

            print(option)

        while True:

            answer = input(
                "\nYour answer (A/B/C/D): "
            ).strip().upper()

            if answer in ["A", "B", "C", "D"]:

                break

            print(
                "Please enter A, B, C or D."
            )

        correct_answer = (
            question_data["answer"]
        )

        if answer == correct_answer:

            score += 1

            correct_answers.append(
                {
                    "Question":
                        question_data["question"],

                    "Your Answer":
                        answer,

                    "Correct Answer":
                        correct_answer
                }
            )

            print("Correct! ✅")

        else:

            wrong_answers.append(
                {
                    "Question":
                        question_data["question"],

                    "Your Answer":
                        answer,

                    "Correct Answer":
                        correct_answer
                }
            )

            print(
                "Wrong! ❌"
            )

            print(
                "Correct answer:",
                correct_answer
            )

    # ========================================================
    # CALCULATE RESULT
    # ========================================================

    end_datetime = datetime.now()

    end_time = time.perf_counter()

    time_taken = (
        end_time - start_time
    )

    percentage = (
        score / number_of_questions
    ) * 100

    accuracy = percentage

    grade = calculate_grade(
        percentage
    )

    date = end_datetime.strftime(
        "%d-%m-%Y"
    )

    completion_time = end_datetime.strftime(
        "%I:%M:%S %p"
    )

    # ========================================================
    # PERFORMANCE LEVEL
    # ========================================================

    if percentage >= 90:

        performance = "Excellent 🌟"

    elif percentage >= 75:

        performance = "Very Good 👍"

    elif percentage >= 60:

        performance = "Good 🙂"

    elif percentage >= 40:

        performance = "Needs Improvement 📚"

    else:

        performance = "Poor - Practice More 💪"

    # ========================================================
    # SAVE RESULT
    # ========================================================

    result = {

        "Student": student_name,

        "Category": category,

        "Score": score,

        "Total Questions":
            number_of_questions,

        "Percentage":
            round(percentage, 2),

        "Accuracy":
            round(accuracy, 2),

        "Grade":
            grade,

        "Time Taken (seconds)":
            round(time_taken, 2),

        "Date":
            date,

        "Completion Time":
            completion_time,

        "Performance":
            performance
    }

    quiz_results.append(
        result
    )

    # ========================================================
    # DISPLAY RESULT
    # ========================================================

    print("\n")

    print("=" * 60)

    print("                    EXAM RESULT")

    print("=" * 60)

    print(
        "Student        :",
        student_name
    )

    print(
        "Category       :",
        category
    )

    print(
        "Score          :",
        score,
        "/",
        number_of_questions
    )

    print(
        "Percentage     :",
        round(percentage, 2),
        "%"
    )

    print(
        "Accuracy       :",
        round(accuracy, 2),
        "%"
    )

    print(
        "Grade          :",
        grade
    )

    print(
        "Time Taken     :",
        f"{time_taken:.2f}",
        "seconds"
    )

    print(
        "Completion Time:",
        completion_time
    )

    print(
        "Performance    :",
        performance
    )

    print("=" * 60)

    # ========================================================
    # ANSWER REVIEW
    # ========================================================

    print("\nANSWER REVIEW")

    print("-" * 60)

    print(
        "Correct Answers:",
        len(correct_answers)
    )

    print(
        "Wrong Answers  :",
        len(wrong_answers)
    )

    if len(wrong_answers) > 0:

        print("\nQuestions to Review:")

        print("-" * 60)

        for item in wrong_answers:

            print(
                "\nQuestion:",
                item["Question"]
            )

            print(
                "Your Answer:",
                item["Your Answer"]
            )

            print(
                "Correct Answer:",
                item["Correct Answer"]
            )

    print("\n" + "=" * 60)

    print(
        "Exam completed successfully! 🎉"
    )

    print("=" * 60)


# ============================================================
# VIEW QUIZ HISTORY
# ============================================================

def view_quiz_history():

    print("\n" + "=" * 60)
    print("                    QUIZ HISTORY")
    print("=" * 60)

    if len(quiz_results) == 0:

        print(
            "No quiz results available."
        )

        return

    df = pd.DataFrame(
        quiz_results
    )

    print(
        df.to_string(
            index=False
        )
    )


# ============================================================
# QUIZ STATISTICS
# ============================================================

def quiz_statistics():

    print("\n" + "=" * 60)
    print("                  QUIZ STATISTICS")
    print("=" * 60)

    if len(quiz_results) == 0:

        print(
            "No quiz results available."
        )

        return

    df = pd.DataFrame(
        quiz_results
    )

    total_attempts = len(df)

    average_score = df[
        "Percentage"
    ].mean()

    highest_score = df[
        "Percentage"
    ].max()

    lowest_score = df[
        "Percentage"
    ].min()

    best_attempt = df.loc[
        df["Percentage"].idxmax()
    ]

    print(
        "Total Attempts  :",
        total_attempts
    )

    print(
        "Average Score   :",
        round(
            average_score,
            2
        ),
        "%"
    )

    print(
        "Highest Score   :",
        highest_score,
        "%"
    )

    print(
        "Lowest Score    :",
        lowest_score,
        "%"
    )

    print(
        "Best Student    :",
        best_attempt["Student"]
    )

    print(
        "Best Category   :",
        best_attempt["Category"]
    )

    print(
        "Best Score      :",
        best_attempt["Percentage"],
        "%"
    )

    print("\nCategory Performance:")

    category_average = (
        df.groupby(
            "Category"
        )["Percentage"]
        .mean()
    )

    for category, average in category_average.items():

        print(
            f"{category:20}: "
            f"{average:.2f}%"
        )


# ============================================================
# SAVE RESULTS
# ============================================================

def save_results():

    if len(quiz_results) == 0:

        print(
            "\nNo results available to save."
        )

        return

    df = pd.DataFrame(
        quiz_results
    )

    df.to_csv(
        RESULT_FILE,
        index=False
    )

    print(
        "\nQuiz results saved successfully!"
    )

    print(
        "File:",
        RESULT_FILE
    )


# ============================================================
# LOAD RESULTS
# ============================================================

def load_results():

    global quiz_results

    if os.path.exists(
        RESULT_FILE
    ):

        try:

            df = pd.read_csv(
                RESULT_FILE
            )

            quiz_results = (
                df.to_dict(
                    "records"
                )
            )

            print(
                f"{len(quiz_results)} "
                "previous result(s) loaded."
            )

        except Exception as error:

            print(
                "Error while loading results:",
                error
            )

    else:

        print(
            "No previous quiz results found."
        )


# ============================================================
# MAIN MENU
# ============================================================

def main_menu():

    while True:

        print("\n")

        print("=" * 60)

        print(
            "             SMART QUIZ & EXAM SYSTEM"
        )

        print("=" * 60)

        print("1.  Start Quiz")

        print("2.  View Quiz History")

        print("3.  View Quiz Statistics")

        print("4.  Save Results")

        print("5.  Load Results")

        print("6.  Exit")

        print("=" * 60)

        choice = input(
            "Enter your choice: "
        ).strip()

        if choice == "1":

            start_quiz()

        elif choice == "2":

            view_quiz_history()

        elif choice == "3":

            quiz_statistics()

        elif choice == "4":

            save_results()

        elif choice == "5":

            load_results()

        elif choice == "6":

            print("\n" + "=" * 60)

            print(
                "Thank you for using "
                "Smart Quiz & Exam System!"
            )

            print(
                "Goodbye! 👋"
            )

            print("=" * 60)

            break

        else:

            print(
                "\nInvalid choice!"
            )

            print(
                "Please enter a number from 1 to 6."
            )


# ============================================================
# START PROJECT
# ============================================================

print("=" * 60)

print(
    "           SMART QUIZ & EXAM SYSTEM"
)

print("=" * 60)

now = datetime.now()

print(
    "Date:",
    now.strftime("%d-%m-%Y")
)

print(
    "Time:",
    now.strftime("%I:%M:%S %p")
)

print("=" * 60)

# Load previous results
load_results()

# Start menu
main_menu()
```
