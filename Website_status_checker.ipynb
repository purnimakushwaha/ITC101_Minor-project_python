```python
# ============================================================
#             WEBSITE STATUS CHECKER
#          & RESPONSE TIME ANALYZER
#
# ============================================================

import requests
import pandas as pd
import os
import time
from datetime import datetime


# ============================================================
# PROJECT SETTINGS
# ============================================================

REPORT_FILE = "website_status_report.csv"

website_data = []


# ============================================================
# DEFAULT WEBSITES
# ============================================================

default_websites = [
    "https://www.google.com",
    "https://www.youtube.com",
    "https://www.wikipedia.org",
    "https://www.github.com",
    "https://www.python.org"
]


# ============================================================
# CLEAN URL
# ============================================================

def clean_url(url):

    url = url.strip()

    if not url.startswith("http://") and not url.startswith("https://"):

        url = "https://" + url

    return url


# ============================================================
# CHECK SINGLE WEBSITE
# ============================================================

def check_website(url):

    url = clean_url(url)

    start_time = time.perf_counter()

    try:

        response = requests.get(
            url,
            timeout=10
        )

        end_time = time.perf_counter()

        response_time = (
            end_time - start_time
        )

        status_code = response.status_code

        if 200 <= status_code < 400:

            status = "ONLINE"

        else:

            status = "ERROR"


        result = {

            "Checked At":
                datetime.now().strftime(
                    "%d-%m-%Y %I:%M:%S %p"
                ),

            "Website":
                url,

            "Status":
                status,

            "Status Code":
                status_code,

            "Response Time (seconds)":
                round(
                    response_time,
                    4
                ),

            "Error":
                ""
        }


        return result


    except requests.exceptions.Timeout:

        end_time = time.perf_counter()

        response_time = (
            end_time - start_time
        )


        return {

            "Checked At":
                datetime.now().strftime(
                    "%d-%m-%Y %I:%M:%S %p"
                ),

            "Website":
                url,

            "Status":
                "TIMEOUT",

            "Status Code":
                "N/A",

            "Response Time (seconds)":
                round(
                    response_time,
                    4
                ),

            "Error":
                "Request timed out"
        }


    except requests.exceptions.ConnectionError:

        end_time = time.perf_counter()

        response_time = (
            end_time - start_time
        )


        return {

            "Checked At":
                datetime.now().strftime(
                    "%d-%m-%Y %I:%M:%S %p"
                ),

            "Website":
                url,

            "Status":
                "OFFLINE",

            "Status Code":
                "N/A",

            "Response Time (seconds)":
                round(
                    response_time,
                    4
                ),

            "Error":
                "Connection error"
        }


    except requests.exceptions.RequestException as error:

        end_time = time.perf_counter()

        response_time = (
            end_time - start_time
        )


        return {

            "Checked At":
                datetime.now().strftime(
                    "%d-%m-%Y %I:%M:%S %p"
                ),

            "Website":
                url,

            "Status":
                "ERROR",

            "Status Code":
                "N/A",

            "Response Time (seconds)":
                round(
                    response_time,
                    4
                ),

            "Error":
                str(error)
        }


# ============================================================
# CHECK ONE WEBSITE
# ============================================================

def check_one_website():

    print("\n" + "=" * 70)

    print(
        "                  CHECK ONE WEBSITE"
    )

    print("=" * 70)


    url = input(
        "Enter website URL: "
    ).strip()


    if url == "":

        print(
            "\nURL cannot be empty."
        )

        return


    print(
        "\nChecking website..."
    )


    result = check_website(
        url
    )


    website_data.append(
        result
    )


    print("\n" + "-" * 70)

    print(
        "Website       :",
        result["Website"]
    )

    print(
        "Status        :",
        result["Status"]
    )

    print(
        "Status Code   :",
        result["Status Code"]
    )

    print(
        "Response Time :",
        result["Response Time (seconds)"],
        "seconds"
    )


    if result["Error"]:

        print(
            "Error         :",
            result["Error"]
        )

    print("-" * 70)


# ============================================================
# CHECK MULTIPLE WEBSITES
# ============================================================

def check_multiple_websites():

    print("\n" + "=" * 70)

    print(
        "                 CHECK MULTIPLE WEBSITES"
    )

    print("=" * 70)

    print(
        "Enter one URL per line."
    )

    print(
        "Type 'done' when finished."
    )


    urls = []


    while True:

        url = input(
            "\nWebsite URL: "
        ).strip()


        if url.lower() == "done":

            break


        if url == "":

            print(
                "URL cannot be empty."
            )

            continue


        urls.append(
            url
        )


    if len(urls) == 0:

        print(
            "\nNo websites entered."
        )

        return


    print("\n" + "=" * 70)

    print(
        "                    CHECKING"
    )

    print("=" * 70)


    for index, url in enumerate(
        urls,
        start=1
    ):

        print(
            f"\n[{index}/{len(urls)}] Checking {url}..."
        )


        result = check_website(
            url
        )


        website_data.append(
            result
        )


        print(
            f"Status: {result['Status']}"
        )

        print(
            f"Response Time: "
            f"{result['Response Time (seconds)']} seconds"
        )


    print(
        "\nAll websites checked successfully."
    )


# ============================================================
# CHECK DEFAULT WEBSITES
# ============================================================

def check_default_websites():

    print("\n" + "=" * 70)

    print(
        "                 CHECKING DEFAULT WEBSITES"
    )

    print("=" * 70)


    for index, url in enumerate(
        default_websites,
        start=1
    ):

        print(
            f"\n[{index}/{len(default_websites)}] "
            f"Checking {url}..."
        )


        result = check_website(
            url
        )


        website_data.append(
            result
        )


        print(
            "Status:",
            result["Status"]
        )

        print(
            "Status Code:",
            result["Status Code"]
        )

        print(
            "Response Time:",
            result[
                "Response Time (seconds)"
            ],
            "seconds"
        )


# ============================================================
# VIEW ALL RESULTS
# ============================================================

def view_results():

    print("\n" + "=" * 70)

    print(
        "                   WEBSITE RESULTS"
    )

    print("=" * 70)


    if len(website_data) == 0:

        print(
            "\nNo website results available."
        )

        return


    for index, result in enumerate(
        website_data,
        start=1
    ):

        print("\n" + "-" * 70)

        print(
            f"Result #{index}"
        )

        print("-" * 70)

        print(
            "Website       :",
            result["Website"]
        )

        print(
            "Status        :",
            result["Status"]
        )

        print(
            "Status Code   :",
            result["Status Code"]
        )

        print(
            "Response Time :",
            result[
                "Response Time (seconds)"
            ],
            "seconds"
        )

        if result["Error"]:

            print(
                "Error         :",
                result["Error"]
            )


# ============================================================
# ANALYZE RESULTS
# ============================================================

def analyze_results():

    print("\n" + "=" * 70)

    print(
        "                 WEBSITE ANALYSIS"
    )

    print("=" * 70)


    if len(website_data) == 0:

        print(
            "\nNo results available."
        )

        return


    start_time = time.perf_counter()


    df = pd.DataFrame(
        website_data
    )


    total_websites = len(df)


    online = len(
        df[
            df["Status"] == "ONLINE"
        ]
    )


    offline = len(
        df[
            df["Status"] == "OFFLINE"
        ]
    )


    timeout = len(
        df[
            df["Status"] == "TIMEOUT"
        ]
    )


    errors = len(
        df[
            df["Status"] == "ERROR"
        ]
    )


    successful_times = df[
        df["Status"] == "ONLINE"
    ]["Response Time (seconds)"]


    if len(successful_times) > 0:

        average_response = (
            successful_times.mean()
        )

        fastest_index = (
            successful_times.idxmin()
        )

        slowest_index = (
            successful_times.idxmax()
        )

        fastest_website = df.loc[
            fastest_index,
            "Website"
        ]

        fastest_time = df.loc[
            fastest_index,
            "Response Time (seconds)"
        ]

        slowest_website = df.loc[
            slowest_index,
            "Website"
        ]

        slowest_time = df.loc[
            slowest_index,
            "Response Time (seconds)"
        ]

    else:

        average_response = 0

        fastest_website = "N/A"

        fastest_time = 0

        slowest_website = "N/A"

        slowest_time = 0


    end_time = time.perf_counter()

    analysis_time = (
        end_time - start_time
    )


    print(
        f"\nTotal Websites Checked : "
        f"{total_websites}"
    )

    print(
        f"Online                  : "
        f"{online}"
    )

    print(
        f"Offline                 : "
        f"{offline}"
    )

    print(
        f"Timeout                 : "
        f"{timeout}"
    )

    print(
        f"Errors                  : "
        f"{errors}"
    )

    print(
        f"Average Response Time   : "
        f"{average_response:.4f} seconds"
    )

    print(
        f"\nFastest Website         : "
        f"{fastest_website}"
    )

    print(
        f"Fastest Response Time   : "
        f"{fastest_time} seconds"
    )

    print(
        f"\nSlowest Website         : "
        f"{slowest_website}"
    )

    print(
        f"Slowest Response Time   : "
        f"{slowest_time} seconds"
    )

    print(
        f"\nAnalysis Execution Time : "
        f"{analysis_time:.6f} seconds"
    )


    if total_websites > 0:

        uptime_percentage = (
            online / total_websites
        ) * 100


        print(
            f"\nOnline Percentage       : "
            f"{uptime_percentage:.2f}%"
        )


    print("\n" + "-" * 70)

    print(
        "                    STATUS SUMMARY"
    )

    print("-" * 70)


    summary = {

        "ONLINE": online,

        "OFFLINE": offline,

        "TIMEOUT": timeout,

        "ERROR": errors
    }


    for status, count in summary.items():

        print(
            f"{status:<10}: {count}"
        )


# ============================================================
# SEARCH WEBSITE
# ============================================================

def search_website():

    print("\n" + "=" * 70)

    print(
        "                    SEARCH WEBSITE"
    )

    print("=" * 70)


    if len(website_data) == 0:

        print(
            "\nNo website data available."
        )

        return


    keyword = input(
        "Enter website name or keyword: "
    ).strip().lower()


    if keyword == "":

        print(
            "Please enter a keyword."
        )

        return


    found = []


    for result in website_data:

        if keyword in result[
            "Website"
        ].lower():

            found.append(
                result
            )


    if len(found) == 0:

        print(
            "\nNo matching website found."
        )

        return


    print(
        f"\n{len(found)} result(s) found:"
    )


    for index, result in enumerate(
        found,
        start=1
    ):

        print("\n" + "-" * 60)

        print(
            f"{index}. {result['Website']}"
        )

        print(
            "Status:",
            result["Status"]
        )

        print(
            "Response:",
            result[
                "Response Time (seconds)"
            ],
            "seconds"
        )


# ============================================================
# FILTER ONLINE WEBSITES
# ============================================================

def show_online_websites():

    print("\n" + "=" * 70)

    print(
        "                   ONLINE WEBSITES"
    )

    print("=" * 70)


    if len(website_data) == 0:

        print(
            "No data available."
        )

        return


    online_sites = [

        result

        for result in website_data

        if result["Status"] == "ONLINE"
    ]


    if len(online_sites) == 0:

        print(
            "No online websites found."
        )

        return


    for index, result in enumerate(
        online_sites,
        start=1
    ):

        print(
            f"{index}. "
            f"{result['Website']} "
            f"→ "
            f"{result['Response Time (seconds)']} sec"
        )


# ============================================================
# SAVE CSV REPORT
# ============================================================

def save_report():

    print("\n" + "=" * 70)

    print(
        "                    SAVE REPORT"
    )

    print("=" * 70)


    if len(website_data) == 0:

        print(
            "No data available to save."
        )

        return


    try:

        df = pd.DataFrame(
            website_data
        )


        df.to_csv(
            REPORT_FILE,
            index=False
        )


        print(
            "\nReport saved successfully!"
        )

        print(
            "File:",
            REPORT_FILE
        )

        print(
            "Location:",
            os.path.abspath(
                REPORT_FILE
            )
        )


    except Exception as error:

        print(
            "Error while saving report:"
        )

        print(
            error
        )


# ============================================================
# LOAD PREVIOUS REPORT
# ============================================================

def load_report():

    global website_data


    print("\n" + "=" * 70)

    print(
        "                    LOAD REPORT"
    )

    print("=" * 70)


    if not os.path.exists(
        REPORT_FILE
    ):

        print(
            "No previous report found."
        )

        return


    try:

        df = pd.read_csv(
            REPORT_FILE
        )


        website_data = (
            df.to_dict(
                orient="records"
            )
        )


        print(
            f"{len(website_data)} "
            "record(s) loaded successfully."
        )


    except Exception as error:

        print(
            "Unable to load report:"
        )

        print(
            error
        )


# ============================================================
# CLEAR CURRENT DATA
# ============================================================

def clear_data():

    global website_data


    if len(website_data) == 0:

        print(
            "\nNo data to clear."
        )

        return


    confirmation = input(
        "\nAre you sure you want to clear "
        "current results? (yes/no): "
    ).strip().lower()


    if confirmation == "yes":

        website_data = []

        print(
            "\nCurrent results cleared."
        )

        print(
            "The CSV report has NOT been deleted."
        )


    else:

        print(
            "\nOperation cancelled."
        )


# ============================================================
# PROJECT INFORMATION
# ============================================================

def project_information():

    print("\n" + "=" * 70)

    print(
        "                  PROJECT INFORMATION"
    )

    print("=" * 70)


    print(
        "\nProject Name:"
    )

    print(
        "Website Status Checker & "
        "Response Time Analyzer"
    )


    print(
        "\nPurpose:"
    )

    print(
        "This project checks whether websites are "
        "responding and measures their response time."
    )


    print(
        "\nMain Python Concepts:"
    )

    print(
        "• HTTP Requests"
    )

    print(
        "• Exception Handling"
    )

    print(
        "• File Handling"
    )

    print(
        "• Pandas"
    )

    print(
        "• CSV"
    )

    print(
        "• Functions"
    )

    print(
        "• Execution Time"
    )


    print(
        "\nStatus Codes:"
    )

    print(
        "200 → Successful request"
    )

    print(
        "301/302 → Redirect"
    )

    print(
        "404 → Page not found"
    )

    print(
        "500 → Server error"
    )


    print(
        "\nImportant:"
    )

    print(
        "A website marked OFFLINE or ERROR by this "
        "program does not necessarily mean the entire "
        "website is permanently down."
    )

    print(
        "It only means that the request from this "
        "program could not successfully complete."
    )


# ============================================================
# MAIN MENU
# ============================================================

def main_menu():

    while True:

        print("\n")

        print("=" * 70)

        print(
            "             WEBSITE STATUS CHECKER"
        )

        print(
            "          RESPONSE TIME ANALYZER"
        )

        print("=" * 70)


        print(
            "1.  Check One Website"
        )

        print(
            "2.  Check Multiple Websites"
        )

        print(
            "3.  Check Default Websites"
        )

        print(
            "4.  View All Results"
        )

        print(
            "5.  Analyze Results"
        )

        print(
            "6.  Search Website"
        )

        print(
            "7.  Show Online Websites"
        )

        print(
            "8.  Save CSV Report"
        )

        print(
            "9.  Load Previous Report"
        )

        print(
            "10. Clear Current Data"
        )

        print(
            "11. Project Information"
        )

        print(
            "12. Exit"
        )


        print("=" * 70)


        choice = input(
            "Enter your choice: "
        ).strip()


        if choice == "1":

            check_one_website()


        elif choice == "2":

            check_multiple_websites()


        elif choice == "3":

            check_default_websites()


        elif choice == "4":

            view_results()


        elif choice == "5":

            analyze_results()


        elif choice == "6":

            search_website()


        elif choice == "7":

            show_online_websites()


        elif choice == "8":

            save_report()


        elif choice == "9":

            load_report()


        elif choice == "10":

            clear_data()


        elif choice == "11":

            project_information()


        elif choice == "12":

            print(
                "\n" + "=" * 70
            )

            print(
                "Thank you for using "
                "Website Status Checker!"
            )


            print(
                "=" * 70
            )

            break


        else:

            print(
                "\nInvalid choice!"
            )

            print(
                "Please enter a number from 1 to 12."
            )


# ============================================================
# START PROJECT
# ============================================================

print("=" * 70)

print(
    "             WEBSITE STATUS CHECKER"
)

print(
    "          & RESPONSE TIME ANALYZER"
)



print("=" * 70)


current_time = datetime.now()


print(
    "Date:",
    current_time.strftime(
        "%d-%m-%Y"
    )
)


print(
    "Time:",
    current_time.strftime(
        "%I:%M:%S %p"
    )
)


print("=" * 70)


print(
    "\nProject started successfully!"
)

print(
    "Choose an option from the menu."
)


main_menu()
```
