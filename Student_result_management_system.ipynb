```python
# ============================================================
#       STUDENT RESULT MANAGEMENT & PERFORMANCE SYSTEM
# ============================================================

import pandas as pd
import os
import time
from datetime import datetime


# ============================================================
# PROJECT DATA
# ============================================================

subjects = [
    "Python",
    "Mathematics",
    "English",
    "Computer",
    "Science"
]

students = []

FILE_NAME = "student_results.csv"


# ============================================================
# GRADE CALCULATION
# ============================================================

def calculate_grade(percentage):

    if percentage >= 90:
        return "A+"
    elif percentage >= 80:
        return "A"
    elif percentage >= 70:
        return "B+"
    elif percentage >= 60:
        return "B"
    elif percentage >= 50:
        return "C"
    elif percentage >= 40:
        return "D"
    else:
        return "F"


# ============================================================
# PASS / FAIL
# ============================================================

def calculate_result(marks):

    for mark in marks.values():

        if mark < 33:
            return "Fail"

    return "Pass"


# ============================================================
# GET CURRENT DATE & TIME
# ============================================================

def get_datetime():

    now = datetime.now()

    date = now.strftime("%d-%m-%Y")

    current_time = now.strftime("%I:%M:%S %p")

    return date, current_time


# ============================================================
# ADD STUDENT
# ============================================================

def add_student():

    start_time = time.perf_counter()

    print("\n" + "=" * 60)
    print("                    ADD NEW STUDENT")
    print("=" * 60)

    name = input("Enter student name: ").strip()

    roll_no = input("Enter roll number: ").strip()

    # Check duplicate roll number
    for student in students:

        if str(student["Roll No"]) == roll_no:

            print("\nThis roll number already exists.")

            return

    marks = {}

    # Enter marks
    for subject in subjects:

        while True:

            try:

                mark = float(
                    input(
                        f"Enter marks for {subject} (0-100): "
                    )
                )

                if 0 <= mark <= 100:

                    marks[subject] = mark

                    break

                else:

                    print(
                        "Marks must be between 0 and 100."
                    )

            except ValueError:

                print(
                    "Please enter a valid number."
                )

    total = sum(marks.values())

    percentage = total / len(subjects)

    grade = calculate_grade(percentage)

    result = calculate_result(marks)

    date, current_time = get_datetime()

    student = {

        "Roll No": roll_no,

        "Name": name,

        "Python": marks["Python"],

        "Mathematics": marks["Mathematics"],

        "English": marks["English"],

        "Computer": marks["Computer"],

        "Science": marks["Science"],

        "Total": total,

        "Percentage": round(percentage, 2),

        "Grade": grade,

        "Result": result,

        "Date": date,

        "Time": current_time
    }

    students.append(student)

    execution_time = time.perf_counter() - start_time

    print("\n" + "-" * 60)

    print("             STUDENT ADDED SUCCESSFULLY")

    print("-" * 60)

    print("Name       :", name)

    print("Roll No    :", roll_no)

    print("Total      :", total)

    print("Percentage :", round(percentage, 2), "%")

    print("Grade      :", grade)

    print("Result     :", result)

    print("Date       :", date)

    print("Time       :", current_time)

    print(
        "Execution Time :",
        f"{execution_time:.6f}",
        "seconds"
    )

    print("-" * 60)


# ============================================================
# VIEW ALL STUDENTS
# ============================================================

def view_students():

    start_time = time.perf_counter()

    print("\n" + "=" * 60)
    print("                  ALL STUDENT RECORDS")
    print("=" * 60)

    if len(students) == 0:

        print("No student records available.")

        return

    df = pd.DataFrame(students)

    print(
        df.to_string(index=False)
    )

    execution_time = time.perf_counter() - start_time

    print(
        "\nExecution Time:",
        f"{execution_time:.6f}",
        "seconds"
    )


# ============================================================
# SEARCH STUDENT
# ============================================================

def search_student():

    start_time = time.perf_counter()

    print("\n" + "=" * 60)
    print("                    SEARCH STUDENT")
    print("=" * 60)

    if len(students) == 0:

        print("No student records available.")

        return

    print("1. Search by Roll Number")
    print("2. Search by Name")

    choice = input(
        "Choose option: "
    ).strip()

    found_students = []

    if choice == "1":

        roll_no = input(
            "Enter roll number: "
        ).strip()

        for student in students:

            if str(student["Roll No"]) == roll_no:

                found_students.append(student)

    elif choice == "2":

        name = input(
            "Enter student name: "
        ).strip().lower()

        for student in students:

            if name in str(
                student["Name"]
            ).lower():

                found_students.append(student)

    else:

        print("Invalid choice.")

        return

    if len(found_students) == 0:

        print("\nStudent not found.")

    else:

        print("\nStudent Found!")

        print("-" * 60)

        for student in found_students:

            for key, value in student.items():

                print(
                    f"{key:15}: {value}"
                )

            print("-" * 60)

    execution_time = time.perf_counter() - start_time

    print(
        "Search Execution Time:",
        f"{execution_time:.6f}",
        "seconds"
    )


# ============================================================
# UPDATE STUDENT MARKS
# ============================================================

def update_student():

    print("\n" + "=" * 60)
    print("                  UPDATE STUDENT MARKS")
    print("=" * 60)

    if len(students) == 0:

        print("No student records available.")

        return

    roll_no = input(
        "Enter roll number: "
    ).strip()

    for student in students:

        if str(student["Roll No"]) == roll_no:

            print(
                "\nStudent:",
                student["Name"]
            )

            print("\nSelect subject to update:")

            for i, subject in enumerate(
                subjects,
                start=1
            ):

                print(
                    f"{i}. {subject}"
                )

            try:

                choice = int(
                    input(
                        "Enter subject number: "
                    )
                )

                if not 1 <= choice <= len(subjects):

                    print("Invalid subject.")

                    return

                subject = subjects[choice - 1]

                while True:

                    try:

                        new_mark = float(
                            input(
                                f"Enter new marks for {subject}: "
                            )
                        )

                        if 0 <= new_mark <= 100:

                            break

                        print(
                            "Marks must be between 0 and 100."
                        )

                    except ValueError:

                        print(
                            "Enter a valid number."
                        )

                student[subject] = new_mark

                marks = {
                    subject: student[subject]
                    for subject in subjects
                }

                total = sum(
                    marks.values()
                )

                percentage = (
                    total / len(subjects)
                )

                student["Total"] = total

                student["Percentage"] = round(
                    percentage,
                    2
                )

                student["Grade"] = calculate_grade(
                    percentage
                )

                student["Result"] = calculate_result(
                    marks
                )

                date, current_time = get_datetime()

                student["Date"] = date

                student["Time"] = current_time

                print(
                    "\nMarks updated successfully!"
                )

                print(
                    "New Percentage:",
                    student["Percentage"],
                    "%"
                )

                print(
                    "New Grade:",
                    student["Grade"]
                )

                print(
                    "New Result:",
                    student["Result"]
                )

                return

            except ValueError:

                print(
                    "Please enter a valid number."
                )

                return

    print("Student not found.")


# ============================================================
# DELETE STUDENT
# ============================================================

def delete_student():

    print("\n" + "=" * 60)
    print("                    DELETE STUDENT")
    print("=" * 60)

    if len(students) == 0:

        print("No student records available.")

        return

    roll_no = input(
        "Enter roll number: "
    ).strip()

    for student in students:

        if str(student["Roll No"]) == roll_no:

            students.remove(student)

            print(
                "\nStudent deleted successfully."
            )

            return

    print("Student not found.")


# ============================================================
# INDIVIDUAL REPORT CARD
# ============================================================

def generate_report_card():

    print("\n" + "=" * 60)
    print("                    STUDENT REPORT CARD")
    print("=" * 60)

    if len(students) == 0:

        print("No student records available.")

        return

    roll_no = input(
        "Enter roll number: "
    ).strip()

    for student in students:

        if str(student["Roll No"]) == roll_no:

            print("\n")
            print("=" * 55)

            print(
                "             STUDENT REPORT CARD"
            )

            print("=" * 55)

            print(
                "Name       :",
                student["Name"]
            )

            print(
                "Roll No    :",
                student["Roll No"]
            )

            print("-" * 55)

            print(
                f"{'Subject':20} {'Marks':>10}"
            )

            print("-" * 55)

            for subject in subjects:

                print(
                    f"{subject:20} "
                    f"{student[subject]:>10}"
                )

            print("-" * 55)

            print(
                "Total      :",
                student["Total"]
            )

            print(
                "Percentage :",
                student["Percentage"],
                "%"
            )

            print(
                "Grade      :",
                student["Grade"]
            )

            print(
                "Result     :",
                student["Result"]
            )

            print("-" * 55)

            print(
                "Date       :",
                student["Date"]
            )

            print(
                "Time       :",
                student["Time"]
            )

            print("=" * 55)

            return

    print("Student not found.")


# ============================================================
# TOPPER
# ============================================================

def find_topper():

    print("\n" + "=" * 60)
    print("                      CLASS TOPPER")
    print("=" * 60)

    if len(students) == 0:

        print("No student records available.")

        return

    df = pd.DataFrame(students)

    topper = df.loc[
        df["Percentage"].idxmax()
    ]

    print(
        "Topper Name       :",
        topper["Name"]
    )

    print(
        "Roll No           :",
        topper["Roll No"]
    )

    print(
        "Percentage        :",
        topper["Percentage"],
        "%"
    )

    print(
        "Grade             :",
        topper["Grade"]
    )


# ============================================================
# PERFORMANCE ANALYSIS
# ============================================================

def performance_analysis():

    start_time = time.perf_counter()

    print("\n" + "=" * 60)
    print("                  PERFORMANCE ANALYSIS")
    print("=" * 60)

    if len(students) == 0:

        print("No student records available.")

        return

    df = pd.DataFrame(students)

    print(
        "\nTotal Students:",
        len(df)
    )

    print(
        "Class Average:",
        round(
            df["Percentage"].mean(),
            2
        ),
        "%"
    )

    print(
        "Highest Percentage:",
        df["Percentage"].max(),
        "%"
    )

    print(
        "Lowest Percentage:",
        df["Percentage"].min(),
        "%"
    )

    # Pass percentage
    passed = len(
        df[df["Result"] == "Pass"]
    )

    pass_percentage = (
        passed / len(df)
    ) * 100

    print(
        "Pass Percentage:",
        round(
            pass_percentage,
            2
        ),
        "%"
    )

    # Subject analysis
    print("\nSubject-wise Average Marks:")

    print("-" * 40)

    subject_averages = {}

    for subject in subjects:

        average = df[subject].mean()

        subject_averages[subject] = average

        print(
            f"{subject:15}: {average:.2f}"
        )

    best_subject = max(
        subject_averages,
        key=subject_averages.get
    )

    weakest_subject = min(
        subject_averages,
        key=subject_averages.get
    )

    print("\nBest Performing Subject:")

    print(
        best_subject,
        "-",
        round(
            subject_averages[best_subject],
            2
        )
    )

    print("\nWeakest Performing Subject:")

    print(
        weakest_subject,
        "-",
        round(
            subject_averages[weakest_subject],
            2
        )
    )

    # Students needing improvement
    print("\nStudents Needing Improvement:")

    weak_students = df[
        df["Percentage"] < 40
    ]

    if len(weak_students) == 0:

        print(
            "No students below 40%."
        )

    else:

        for _, student in weak_students.iterrows():

            print(
                f"- {student['Name']} "
                f"({student['Percentage']}%)"
            )

    # Grade distribution
    print("\nGrade Distribution:")

    grade_counts = df[
        "Grade"
    ].value_counts()

    for grade, count in grade_counts.items():

        print(
            f"Grade {grade}: {count} student(s)"
        )

    execution_time = (
        time.perf_counter()
        - start_time
    )

    print(
        "\nAnalysis Execution Time:",
        f"{execution_time:.6f}",
        "seconds"
    )


# ============================================================
# SAVE DATA
# ============================================================

def save_data():

    if len(students) == 0:

        print(
            "\nNo student data available."
        )

        return

    df = pd.DataFrame(students)

    df.to_csv(
        FILE_NAME,
        index=False
    )

    print(
        f"\nData saved successfully!"
    )

    print(
        "File:",
        FILE_NAME
    )


# ============================================================
# LOAD DATA
# ============================================================

def load_data():

    global students

    if os.path.exists(FILE_NAME):

        try:

            df = pd.read_csv(
                FILE_NAME
            )

            students = df.to_dict(
                "records"
            )

            print(
                f"{len(students)} student record(s) loaded."
            )

        except Exception as error:

            print(
                "Error while loading data:",
                error
            )

    else:

        print(
            "No previous data found."
        )


# ============================================================
# MAIN MENU
# ============================================================

def main_menu():

    while True:

        print("\n")

        print("=" * 60)

        print(
            "       STUDENT RESULT MANAGEMENT SYSTEM"
        )

        print("=" * 60)

        print("1.  Add Student")

        print("2.  View All Students")

        print("3.  Search Student")

        print("4.  Update Student Marks")

        print("5.  Delete Student")

        print("6.  Generate Report Card")

        print("7.  Find Class Topper")

        print("8.  Performance Analysis")

        print("9.  Save Data")

        print("10. Load Data")

        print("11. Exit")

        print("=" * 60)

        choice = input(
            "Enter your choice: "
        ).strip()

        if choice == "1":

            add_student()

        elif choice == "2":

            view_students()

        elif choice == "3":

            search_student()

        elif choice == "4":

            update_student()

        elif choice == "5":

            delete_student()

        elif choice == "6":

            generate_report_card()

        elif choice == "7":

            find_topper()

        elif choice == "8":

            performance_analysis()

        elif choice == "9":

            save_data()

        elif choice == "10":

            load_data()

        elif choice == "11":

            print("\n" + "=" * 60)

            print(
                "Thank you for using the system!"
            )

            print(
                "Goodbye!"
            )

            print("=" * 60)

            break

        else:

            print(
                "\nInvalid choice!"
            )

            print(
                "Please select a number from 1 to 11."
            )


# ============================================================
# START PROJECT
# ============================================================

print("=" * 60)

print(
    "       STUDENT RESULT MANAGEMENT SYSTEM"
)

print("=" * 60)

print(
    "Project Started Successfully!"
)

date, current_time = get_datetime()

print(
    "Date:",
    date
)

print(
    "Time:",
    current_time
)

print("=" * 60)

# Load old data if available
load_data()

# Start menu
main_menu()
```
