```python
# ============================================================
#          VEHICLE PARKING MANAGEMENT SYSTEM
# ============================================================

import pandas as pd
import os
import time
from datetime import datetime


# ============================================================
# PARKING SETTINGS
# ============================================================

TOTAL_SLOTS = 10

FILE_NAME = "parking_records.csv"

parking_records = []

# Parking rates per hour
RATES = {
    "Bike": 20,
    "Car": 40,
    "Bus": 80
}


# ============================================================
# DATE AND TIME FUNCTION
# ============================================================

def get_datetime():

    now = datetime.now()

    date = now.strftime("%d-%m-%Y")

    current_time = now.strftime("%I:%M:%S %p")

    return date, current_time


# ============================================================
# FIND AVAILABLE SLOT
# ============================================================

def find_available_slot():

    occupied_slots = []

    for record in parking_records:

        if record["Status"] == "Parked":

            occupied_slots.append(
                int(record["Slot"])
            )

    for slot in range(1, TOTAL_SLOTS + 1):

        if slot not in occupied_slots:

            return slot

    return None


# ============================================================
# VEHICLE ENTRY
# ============================================================

def vehicle_entry():

    start_time = time.perf_counter()

    print("\n" + "=" * 60)
    print("                    VEHICLE ENTRY")
    print("=" * 60)

    # Check parking availability
    slot = find_available_slot()

    if slot is None:

        print("\nParking is FULL!")

        return

    vehicle_number = input(
        "Enter vehicle number: "
    ).strip().upper()

    # Check whether vehicle is already parked
    for record in parking_records:

        if (
            record["Vehicle Number"] == vehicle_number
            and record["Status"] == "Parked"
        ):

            print(
                "\nThis vehicle is already parked."
            )

            return

    print("\nVehicle Type")

    print("1. Bike")

    print("2. Car")

    print("3. Bus")

    choice = input(
        "Choose vehicle type: "
    ).strip()

    if choice == "1":

        vehicle_type = "Bike"

    elif choice == "2":

        vehicle_type = "Car"

    elif choice == "3":

        vehicle_type = "Bus"

    else:

        print("Invalid vehicle type.")

        return

    date, entry_time = get_datetime()

    # Store exact datetime for calculation
    entry_datetime = datetime.now()

    record = {

        "Vehicle Number": vehicle_number,

        "Vehicle Type": vehicle_type,

        "Slot": slot,

        "Entry Date": date,

        "Entry Time": entry_time,

        "Entry Datetime": entry_datetime.strftime(
            "%Y-%m-%d %H:%M:%S"
        ),

        "Exit Date": "",

        "Exit Time": "",

        "Duration": "",

        "Fee": 0,

        "Status": "Parked"
    }

    parking_records.append(record)

    execution_time = (
        time.perf_counter() - start_time
    )

    print("\n" + "-" * 60)

    print("             VEHICLE PARKED SUCCESSFULLY")

    print("-" * 60)

    print(
        "Vehicle Number :",
        vehicle_number
    )

    print(
        "Vehicle Type   :",
        vehicle_type
    )

    print(
        "Parking Slot   :",
        slot
    )

    print(
        "Entry Date     :",
        date
    )

    print(
        "Entry Time     :",
        entry_time
    )

    print(
        "Parking Rate   :",
        f"₹{RATES[vehicle_type]}/hour"
    )

    print(
        "Execution Time :",
        f"{execution_time:.6f}",
        "seconds"
    )

    print("-" * 60)


# ============================================================
# VEHICLE EXIT
# ============================================================

def vehicle_exit():

    start_time = time.perf_counter()

    print("\n" + "=" * 60)
    print("                    VEHICLE EXIT")
    print("=" * 60)

    if len(parking_records) == 0:

        print("No vehicles found.")

        return

    vehicle_number = input(
        "Enter vehicle number: "
    ).strip().upper()

    for record in parking_records:

        if (
            record["Vehicle Number"] == vehicle_number
            and record["Status"] == "Parked"
        ):

            exit_datetime = datetime.now()

            entry_datetime = datetime.strptime(
                record["Entry Datetime"],
                "%Y-%m-%d %H:%M:%S"
            )

            # Calculate duration in seconds
            duration_seconds = (
                exit_datetime - entry_datetime
            ).total_seconds()

            # Convert to hours
            duration_hours = (
                duration_seconds / 3600
            )

            # Minimum 1 hour charge
            charged_hours = max(
                1,
                int(duration_hours) +
                (1 if duration_hours % 1 > 0 else 0)
            )

            rate = RATES[
                record["Vehicle Type"]
            ]

            fee = charged_hours * rate

            exit_date = exit_datetime.strftime(
                "%d-%m-%Y"
            )

            exit_time = exit_datetime.strftime(
                "%I:%M:%S %p"
            )

            # Update record
            record["Exit Date"] = exit_date

            record["Exit Time"] = exit_time

            record["Duration"] = (
                f"{duration_hours:.2f} hours"
            )

            record["Fee"] = fee

            record["Status"] = "Exited"

            execution_time = (
                time.perf_counter()
                - start_time
            )

            print("\n" + "-" * 60)

            print(
                "              VEHICLE EXIT SUCCESSFUL"
            )

            print("-" * 60)

            print(
                "Vehicle Number :",
                record["Vehicle Number"]
            )

            print(
                "Vehicle Type   :",
                record["Vehicle Type"]
            )

            print(
                "Parking Slot   :",
                record["Slot"]
            )

            print(
                "Entry Time     :",
                record["Entry Time"]
            )

            print(
                "Exit Time      :",
                exit_time
            )

            print(
                "Parking Duration:",
                f"{duration_hours:.2f}",
                "hours"
            )

            print(
                "Charged Hours  :",
                charged_hours
            )

            print(
                "Parking Fee    :",
                f"₹{fee}"
            )

            print(
                "Execution Time :",
                f"{execution_time:.6f}",
                "seconds"
            )

            print("-" * 60)

            return

    print(
        "\nVehicle not found in active parking."
    )


# ============================================================
# VIEW PARKED VEHICLES
# ============================================================

def view_parked_vehicles():

    start_time = time.perf_counter()

    print("\n" + "=" * 60)
    print("                   PARKED VEHICLES")
    print("=" * 60)

    parked = []

    for record in parking_records:

        if record["Status"] == "Parked":

            parked.append(record)

    if len(parked) == 0:

        print(
            "No vehicles are currently parked."
        )

        return

    df = pd.DataFrame(parked)

    columns = [
        "Vehicle Number",
        "Vehicle Type",
        "Slot",
        "Entry Date",
        "Entry Time",
        "Status"
    ]

    print(
        df[columns].to_string(
            index=False
        )
    )

    execution_time = (
        time.perf_counter() - start_time
    )

    print(
        "\nExecution Time:",
        f"{execution_time:.6f}",
        "seconds"
    )


# ============================================================
# VIEW ALL PARKING RECORDS
# ============================================================

def view_all_records():

    print("\n" + "=" * 60)
    print("                   ALL PARKING RECORDS")
    print("=" * 60)

    if len(parking_records) == 0:

        print(
            "No parking records available."
        )

        return

    df = pd.DataFrame(
        parking_records
    )

    columns = [
        "Vehicle Number",
        "Vehicle Type",
        "Slot",
        "Entry Date",
        "Entry Time",
        "Exit Date",
        "Exit Time",
        "Duration",
        "Fee",
        "Status"
    ]

    print(
        df[columns].to_string(
            index=False
        )
    )


# ============================================================
# SEARCH VEHICLE
# ============================================================

def search_vehicle():

    start_time = time.perf_counter()

    print("\n" + "=" * 60)
    print("                    SEARCH VEHICLE")
    print("=" * 60)

    if len(parking_records) == 0:

        print(
            "No parking records available."
        )

        return

    vehicle_number = input(
        "Enter vehicle number: "
    ).strip().upper()

    found = False

    for record in parking_records:

        if record["Vehicle Number"] == vehicle_number:

            found = True

            print("\nVehicle Found!")

            print("-" * 45)

            print(
                "Vehicle Number :",
                record["Vehicle Number"]
            )

            print(
                "Vehicle Type   :",
                record["Vehicle Type"]
            )

            print(
                "Slot           :",
                record["Slot"]
            )

            print(
                "Entry Date     :",
                record["Entry Date"]
            )

            print(
                "Entry Time     :",
                record["Entry Time"]
            )

            print(
                "Exit Date      :",
                record["Exit Date"]
            )

            print(
                "Exit Time      :",
                record["Exit Time"]
            )

            print(
                "Duration       :",
                record["Duration"]
            )

            print(
                "Fee            :",
                f"₹{record['Fee']}"
            )

            print(
                "Status         :",
                record["Status"]
            )

            print("-" * 45)

            break

    if not found:

        print(
            "\nVehicle record not found."
        )

    execution_time = (
        time.perf_counter()
        - start_time
    )

    print(
        "Search Execution Time:",
        f"{execution_time:.6f}",
        "seconds"
    )


# ============================================================
# PARKING SLOT STATUS
# ============================================================

def parking_slot_status():

    print("\n" + "=" * 60)
    print("                   PARKING SLOT STATUS")
    print("=" * 60)

    occupied_slots = {}

    for record in parking_records:

        if record["Status"] == "Parked":

            occupied_slots[
                int(record["Slot"])
            ] = record["Vehicle Number"]

    for slot in range(
        1,
        TOTAL_SLOTS + 1
    ):

        if slot in occupied_slots:

            print(
                f"Slot {slot:2} : OCCUPIED "
                f"({occupied_slots[slot]})"
            )

        else:

            print(
                f"Slot {slot:2} : AVAILABLE"
            )


# ============================================================
# PARKING STATISTICS
# ============================================================

def parking_statistics():

    start_time = time.perf_counter()

    print("\n" + "=" * 60)
    print("                    PARKING STATISTICS")
    print("=" * 60)

    total_records = len(
        parking_records
    )

    if total_records == 0:

        print(
            "No parking records available."
        )

        return

    parked_count = sum(
        1
        for record in parking_records
        if record["Status"] == "Parked"
    )

    exited_count = sum(
        1
        for record in parking_records
        if record["Status"] == "Exited"
    )

    total_collection = sum(
        float(record["Fee"])
        for record in parking_records
    )

    available_slots = (
        TOTAL_SLOTS - parked_count
    )

    print(
        "Total Parking Slots :",
        TOTAL_SLOTS
    )

    print(
        "Occupied Slots      :",
        parked_count
    )

    print(
        "Available Slots     :",
        available_slots
    )

    print(
        "Total Vehicles      :",
        total_records
    )

    print(
        "Currently Parked    :",
        parked_count
    )

    print(
        "Vehicles Exited     :",
        exited_count
    )

    print(
        "Total Collection    :",
        f"₹{total_collection:.2f}"
    )

    # Vehicle type statistics
    type_counts = {}

    for record in parking_records:

        vehicle_type = record[
            "Vehicle Type"
        ]

        type_counts[vehicle_type] = (
            type_counts.get(
                vehicle_type,
                0
            ) + 1
        )

    print("\nVehicle Type Statistics:")

    for vehicle_type, count in type_counts.items():

        print(
            f"{vehicle_type}: {count}"
        )

    execution_time = (
        time.perf_counter()
        - start_time
    )

    print(
        "\nAnalysis Execution Time:",
        f"{execution_time:.6f}",
        "seconds"
    )


# ============================================================
# SAVE DATA
# ============================================================

def save_data():

    if len(parking_records) == 0:

        print(
            "\nNo parking data available."
        )

        return

    df = pd.DataFrame(
        parking_records
    )

    df.to_csv(
        FILE_NAME,
        index=False
    )

    print(
        f"\nParking data saved successfully!"
    )

    print(
        "File:",
        FILE_NAME
    )


# ============================================================
# LOAD DATA
# ============================================================

def load_data():

    global parking_records

    if os.path.exists(FILE_NAME):

        try:

            df = pd.read_csv(
                FILE_NAME
            )

            parking_records = (
                df.to_dict(
                    "records"
                )
            )

            print(
                f"{len(parking_records)} "
                "parking record(s) loaded."
            )

        except Exception as error:

            print(
                "Error while loading data:",
                error
            )

    else:

        print(
            "No previous parking data found."
        )


# ============================================================
# MAIN MENU
# ============================================================

def main_menu():

    while True:

        print("\n")

        print("=" * 60)

        print(
            "             VEHICLE PARKING MANAGEMENT"
        )

        print("=" * 60)

        print("1.  Vehicle Entry")

        print("2.  Vehicle Exit")

        print("3.  View Parked Vehicles")

        print("4.  View All Parking Records")

        print("5.  Search Vehicle")

        print("6.  Parking Slot Status")

        print("7.  Parking Statistics")

        print("8.  Save Data")

        print("9.  Load Data")

        print("10. Exit")

        print("=" * 60)

        choice = input(
            "Enter your choice: "
        ).strip()

        if choice == "1":

            vehicle_entry()

        elif choice == "2":

            vehicle_exit()

        elif choice == "3":

            view_parked_vehicles()

        elif choice == "4":

            view_all_records()

        elif choice == "5":

            search_vehicle()

        elif choice == "6":

            parking_slot_status()

        elif choice == "7":

            parking_statistics()

        elif choice == "8":

            save_data()

        elif choice == "9":

            load_data()

        elif choice == "10":

            print("\n" + "=" * 60)

            print(
                "Thank you for using the "
                "Vehicle Parking Management System!"
            )

            print(
                "Goodbye!"
            )

            print("=" * 60)

            break

        else:

            print(
                "\nInvalid choice!"
            )

            print(
                "Please enter a number from 1 to 10."
            )


# ============================================================
# START PROJECT
# ============================================================

print("=" * 60)

print(
    "       VEHICLE PARKING MANAGEMENT SYSTEM"
)

print("=" * 60)

date, current_time = get_datetime()

print(
    "Date:",
    date
)

print(
    "Time:",
    current_time
)

print(
    "Total Parking Slots:",
    TOTAL_SLOTS
)

print("=" * 60)

# Load old records if available
load_data()

# Start application
main_menu()
```
